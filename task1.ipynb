```python
import ast
```

```python
def extract_names (col):
    L=[]
    for i in ast.literal_eval(col):
        L.append(i['name'])
    return " ".join(L)
```

```python
def extract_cast_names(col):
    L=[]
    for i in ast.literal_eval(col):
        L.append(i['name'])
    return L
```

```python
def extract_director(col):
    L=[]
    for i in ast.literal_eval(col):
        if i['job']== 'Director':
            L.append(i['name'])
    return L
```

```python
movie_list['genres'] = movie_list['genres'].apply(extract_names)
movie_list['keywords'] = movie_list['keywords'].apply(extract_names)
movie_list['cast'] =movie_list['cast'].apply(extract_cast_names)
movie_list['production_companies'] = movie_list['production_companies'].apply(extract_names)
movie_list['crew'] = movie_list['crew'].apply(extract_director)
```

```python
movie_list.head(2)
```

```python
movie_list['crew']=movie_list['crew'].apply(lambda x: ''.join(x))
movie_list['cast']=movie_list['cast'].apply(lambda x: ' '.join(x))
```

```python
movie_list.head(2)
```

```python
movie_list['tags'] = movie_list['genres']+" "+movie_list['keywords']+ " " + movie_list['overview'] +" "+ movie_list['production_companies']+" "+movie_list['tagline']+" " +movie_list['cast']+" " + movie_list['crew']
```

```python
movie_list=movie_list.drop(columns=[ 'genres', 'keywords', 'overview',
       'production_companies', 'tagline',
       'cast', 'crew'])
```

```python
import re
```

lets remove puctuation

```python
def remove_puncuation(text):
    return re.sub(r'[^\w\s]',"",text)
```

```python
movie_list['tags']=movie_list['tags'].apply(remove_puncuation)
```

```python
movie_list['tags']
```

now lets remove stop words

```python
from nltk.corpus import stopwords
```

```python
stop =set(stopwords.words('english'))
```

```python
stop
```

```python
movie_list.head(2)
```

```python
def remove_stopword(text):
    li =[]
    for word in text.split():
        if word not in stop:
            li.append(word)
    return " ".join(li)
```

```python
movie_list['tags']=movie_list['tags'].apply(remove_stopword)
```
